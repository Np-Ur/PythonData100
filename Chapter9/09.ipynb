# Menu.84 グラフ作成も生成AIにお任せ！美しい可視化を実現しよう

生成された可視化コード
(japanize_matplotlibのインポートを追記する)


```python
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

# CSV読み込み
df = pd.read_csv("sales_data.csv")

# 日付をdatetime型に変換
df["order_date"] = pd.to_datetime(df["order_date"])

# -----------------------------
# 集計データ作成
# -----------------------------
# 1. カテゴリ別売上合計

category_sales = df.groupby(
"category")["total_amount"].sum().sort_values(ascending=False)

# 2. 月別売上推移
df["year_month"] = df["order_date"].dt.to_period("M").astype(str)
monthly_sales = df.groupby("year_month")["total_amount"].sum()

# 3. 売上金額分布（total_amount をそのまま使用）
sales_amount = df["total_amount"]

# -----------------------------
# グラフ作成
# -----------------------------
plt.figure(figsize=(18, 5))

# (1) カテゴリ別売上合計（棒グラフ）
plt.subplot(1, 3, 1)
category_sales.plot(kind="bar")
plt.title("カテゴリ別売上合計")
plt.xlabel("カテゴリ")
plt.ylabel("売上金額")
plt.xticks(rotation=45, ha="right")

# (2) 月別売上推移（折れ線グラフ）
plt.subplot(1, 3, 2)
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")
plt.title("月別売上推移")
plt.xlabel("年月")
plt.ylabel("売上金額")
plt.xticks(rotation=45)

# (3) 売上金額の分布（ヒストグラム）
plt.subplot(1, 3, 3)
plt.hist(sales_amount, bins=30)
plt.title("売上金額の分布")
plt.xlabel("売上金額")
plt.ylabel("件数")

# レイアウト調整
plt.tight_layout()

# 画像として保存
plt.savefig("sales_visualization.png")

# 表示（不要ならコメントアウト可）
plt.show()
```

# Menu.88 異常値を見逃さない！外れ値&重複データを処理しよう

外れ値を検出するコード

```python
import pandas as pd

df = pd.read_csv("sales_data_cleaned.csv")

NUMERIC_COLUMNS = ["quantity", "total_amount"]
outlier_summary = {}

for col in NUMERIC_COLUMNS:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1

  lower_bound = q1 - 1.5 * iqr
  upper_bound = q3 + 1.5 * iqr

  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

  outlier_summary[col] = {
    "外れ値件数": len(outliers),
    "下限値": lower_bound,
    "上限値": upper_bound,
  }

  print(f"\n■ 列名: {col}")
  print(f"下限値: {lower_bound:.2f}")
  print(f"上限値: {upper_bound:.2f}")
  print(f"外れ値件数: {len(outliers)}")

print("\n=== 重複データ検出 ===")

duplicate_count = df.duplicated().sum()
print(f"重複行数: {duplicate_count}")
```

外れ値と重複データに対処するコード

```python
import pandas as pd

INPUT_FILE = "sales_data_cleaned.csv"
OUTPUT_FILE = "sales_data_final.csv"
NUMERIC_COLUMNS = ["quantity", "total_amount"]

# 1. データの読み込み
df = pd.read_csv(INPUT_FILE)

# 2. 外れ値フラグ付与 初期値として False を設定
df["is_outlier"] = False

for col in NUMERIC_COLUMNS:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1

  lower_bound = q1 - 1.5 * iqr
  upper_bound = q3 + 1.5 * iqr

  # 外れ値条件
  is_outlier_col = (df[col] < lower_bound) | (df[col] > upper_bound)

  # いずれかの数値列で外れ値なら True
  df.loc[is_outlier_col, "is_outlier"] = True

# 3. 重複行の削除（最初の1行を残す）
df = df.drop_duplicates(keep="first")
df.to_csv(OUTPUT_FILE, index=False)
```
