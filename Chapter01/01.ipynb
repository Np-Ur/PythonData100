# Menu.10 必要なデータだけを取り出そう

売上データの数量（quantity）だけを取り出す

```python
import pandas as pd
sales_df = pd.read_csv("sales_data.csv")

# quantity(数量)列だけを取り出す
quantity_col = sales_df["quantity"]
print(quantity_col.head())
```

売上データの購入日と売上合計金額を取り出す

```python
selected_cols = sales_df[["purchase_date", "total_amount"]]
print(selected_cols.head())
```

指定した行だけを取り出し、表示する

```python
# 10行目から15行目のデータ(インデックス9 ～ 14)
iloc_result = sales_df.iloc[9:15]
print(iloc_result)
```

指定した列だけを取り出し、表示する

```python
# 最初の5行の、位置番号2~4の列
iloc_columns_result = sales_df.iloc[:5, 2:5]
print(iloc_columns_result)
```

商品名と売上合計金額を取り出す

```python
# 商品名と売上合計金額の列だけを取り出す
sales_df_loc = sales_df.loc[:, ["product_name", "total_amount"]]

loc_result = sales_df_loc.head()
print(loc_result)
```

条件に合う行を取り出す

```python
# スマートフォンの売上データを取り出す
smartphone_sales = sales_df.loc[
    sales_df["product_category"] == "スマートフォン"
]
print(smartphone_sales.head())
print("スマートフォンのデータ件数", len(smartphone_sales))
```

条件式の結果を確認する

```python
# 条件式の結果を確認する
mask = sales_df["product_category"] == "スマートフォン"

print(mask.head())
```

特定の行をAND条件で取り出す

```python
# 条件を指定
condition = (
    (sales_df["product_category"] == "タブレット") &
    (sales_df["quantity"] >= 3)
)
tablet_bulk = sales_df.loc[condition]

tablet_bulk_head = tablet_bulk.head()
print(tablet_bulk_head)
```

特定の行をOR条件で取り出す

```python
# OR条件:スマートフォンまたはタブレット
condition = (
    (sales_df["product_category"] == "スマートフォン") |
    (sales_df["product_category"] == "タブレット")
)

smartphone_or_tablet = sales_df.loc[condition]
print("スマートフォンまたはタブレットのデータ件数", len(smartphone_or_tablet))
```

# Menu.12 データを並べ替えて傾向を見よう

売上合計金額を並べ替えて、上位5を表示する

```python
import pandas as pd

sales_df = pd.read_csv("sales_data.csv")

# 売上合計金額の降順(高い順)で並べ替え
top_sales = sales_df.sort_values("total_amount", ascending=False)

top_sales_display = top_sales.head()
print("売上合計金額上位5件")
print(top_sales_display)
```

売上合計金額を並べ替えて、下位5を表示する

```python
# 売上合計金額の昇順(安い順)で並べ替え
bottom_sales = sales_df.sort_values("total_amount", ascending=True)

bottom_sales_display = bottom_sales.head()
print("売上合計金額下位5件")
print(bottom_sales_display)
```
