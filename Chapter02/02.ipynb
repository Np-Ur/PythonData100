# Menu.19 結合方法を使い分けよう

# Menu.20 データをグループごとに集計しよう

店舗別の売上合計を集計する

```python
import pandas as pd

sales_df = pd.read_csv("sales_data.csv")
store_master = pd.read_csv("store_master.csv")

# 店舗IDで結合
sales_with_store = pd.merge(sales_df, store_master, on="store_id")

# 店舗名(store_name)ごとに、売上金額(amount)を合計(sum())する
store_sales = sales_with_store.groupby("store_name")["amount"].sum()

print("店舗別売上合計")
print(store_sales.head())
```

さまざまな集計をする

```python
# 店舗別のさまざまな集計
print("店舗別のさまざまな集計")

# 平均値を計算
store_mean = sales_with_store.groupby("store_name")["amount"].mean()
print("店舗ごとの平均売上", store_mean.head(3))

# 件数をカウント
store_count = sales_with_store.groupby("store_name")["amount"].count()
print("店舗ごとの取引件数", store_count.head(3))

# 最大値･最小値
store_max = sales_with_store.groupby("store_name")["amount"].max()
store_min = sales_with_store.groupby("store_name")["amount"].min()
print("店舗ごとの最高売上", store_max.head(3))
print("店舗ごとの最低売上", store_min.head(3))
```

複数の集計を一度に求める

```python
# 店舗ごとに、合計･平均･件数をまとめて算出
store_stats = sales_with_store.groupby("store_name")["amount"].agg(["sum","mean", "count"])
print(store_stats.head(3))
```

# Menu.21 groupby()メソッドでさまざまな集計をしてみよう

商品カテゴリごとの売上を集計する

```python
import pandas as pd

sales_df = pd.read_csv("sales_data.csv")
store_master = pd.read_csv("store_master.csv")
product_master = pd.read_csv("product_master.csv")

sales_with_store = pd.merge(sales_df, store_master, on="store_id") # 店舗IDで結合
sales_with_store_product = pd.merge(sales_with_store, product_master,on="product_id")

# カテゴリごとの売上集計
category_sales = sales_with_store_product.groupby("category")["amount"].sum()

print("カテゴリ別売上")
print(category_sales.sort_values())
```

曜日ごとの売上を集計する

```python
# 曜日情報を追加（0=月, 6=日）
sales_with_store_product["purchase_datetime"] = pd.to_datetime(sales_with_store_product["purchase_datetime"])
sales_with_store_product["weekday"] = sales_with_store_product["purchase_datetime"].dt.dayofweek

# 曜日名に変換するための辞書
weekday_map = {0: "月", 1: "火", 2: "水", 3: "木", 4: "金", 5: "土", 6: "日"}
sales_with_store_product["weekday_name"] = sales_with_store_product["weekday"].map(weekday_map)

# 曜日ごとの売上を集計
weekday_sales = sales_with_store_product.groupby("weekday_name")["amount"].sum()
print("曜日別売上")
print(weekday_sales.sort_values())
```

# Menu.22 ピボットテーブルでクロス集計に挑戦しよう

ピボットテーブルを作成する

```python
import pandas as pd

sales_df = pd.read_csv("sales_data.csv")
store_master = pd.read_csv("store_master.csv")
product_master = pd.read_csv("product_master.csv")

sales_with_store = pd.merge(sales_df, store_master, on="store_id")
sales_with_store_product = pd.merge(sales_with_store, product_master,on="product_id")

# 店舗×カテゴリのクロス集計
pivot_sales = pd.pivot_table(
    sales_with_store_product,
    index="store_name", # 行に店舗名
    columns="category", # 列にカテゴリ
    values="amount", # 集計するのは金額
    aggfunc="sum", # 合計を計算
    fill_value=0 # 売上がない場所は0で埋める
)

print("店舗×カテゴリ別売上")
print(pivot_sales.head())
```

# Menu.29 折れ線グラフで推移を追ってみよう

折れ線グラフを描画する

```python
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

full_sales_df = pd.read_csv("analyzed_sales_data.csv")

# 時間帯ごとの来客数を集計
hourly_customers = full_sales_df.groupby("hour")["transaction_id"].count()

# 折れ線グラフの作成
plt.figure(figsize=(10, 6))
plt.plot(hourly_customers.index, hourly_customers.values, marker="o",linewidth=2, markersize=8)

plt.xlabel("時間帯")
plt.ylabel("来客数")
plt.title("1日の来客数推移")
plt.grid(True, alpha=0.5)
plt.xticks(range(6, 23))
plt.show()
```

グラフに複数の線を重ねる

```python
# 時間帯×カテゴリの売上集計

category_hourly = pd.pivot_table(
    full_sales_df,
    index="hour",
    columns="category",
    values="amount",
    aggfunc="sum"
)

plt.figure(figsize=(12, 6))

# カテゴリごとに線を引く
for category in category_hourly.columns:
  plt.plot(category_hourly.index, category_hourly[category], marker="o",label=category)

plt.xlabel("時間帯")
plt.ylabel("売上(円)")
plt.title("カテゴリ別時間帯売上推移")
plt.legend() # 凡例を表示
plt.grid(True, alpha=0.5)
plt.xticks(range(6, 23))

plt.show()
```

# Menu.30 散布図で変数の関係を発見しよう

散布図を描画する

```python
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

full_sales_df = pd.read_csv("analyzed_sales_data.csv")

# 店舗ごとの集計
store_counts = full_sales_df.groupby("store_name")["transaction_id"].count()
store_sales = full_sales_df.groupby("store_name")["amount"].sum()

# 散布図の作成
plt.figure(figsize=(10, 6))
plt.scatter(store_counts, store_sales, s=100, alpha=0.7, color="blue")
plt.xlabel("取引数")
plt.ylabel("総売上(円)")
plt.title("店舗別:取引数と売上の関係")
plt.grid(True, alpha=0.3)

plt.show()
```
